```python
import numpy as np 
from scipy import stats
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]

sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)

n = len(sample)
mean_population = 170
```

```python
t_test = (sample_mean - mean_population) / (sample_std / np.sqrt(n))
print(t_test)
```

# Using p value 

```python
p_value = 2 * (1-stats.t.cdf(abs(t_test),df = n - 1))
alpha = 0.05
```

```python
if p_value < alpha:
    print("REject the null hypothesis & Accept the Alternate")
else:
    print("Accept the null hypothesis & Reject the Alternate")
```
